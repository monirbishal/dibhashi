Dataset-1

```python
import pandas as pd
import json

# 1️⃣ Read the TSV file (place it in sample_data folder)
df = pd.read_csv('/content/Sentence-pairs-in-English-Bengali.tsv', sep='\t', header=None, usecols=[1, 3], names=['en', 'bn'])

# Inspect first few rows
print(df.head())

# 2️⃣ Drop rows with missing or empty values
df = df.dropna(subset=['en','bn'])
df = df[(df.en.str.strip()!='') & (df.bn.str.strip()!='')]

# 3️⃣ Optionally remove duplicates
df = df.drop_duplicates()

# 4️⃣ Convert to list of dicts
pairs = df.to_dict(orient='records')

# 5️⃣ Save as JSON to sample_data/output.json
with open('sample_data/output.json', 'w', encoding='utf-8') as f:
    json.dump(pairs, f, ensure_ascii=False, indent=2)

print(f"✅ Exported {len(pairs)} pairs to sample_data/output.json")
```
